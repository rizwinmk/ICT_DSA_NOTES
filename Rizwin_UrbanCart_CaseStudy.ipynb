```python
def get_product_price(product_id, products_list):
    for product in products_list:
        if product["Product_ID"] == product_id:
            return product["Price"]

    return None
```

```python
def get_customer_name(customer_id, customers_list):
    for customer in customers_list:
        if customer["Customer_ID"] == customer_id:
            return customer["Name"]
```

```python
def calculate_order_value(price, quantity):
    return price * quantity
```

```python
def classify_order(value, high_limit=3000, low_limit=1000):

    if value > high_limit:
        return "High"

    elif value >= low_limit:
        return "Medium"

    else:
        return "Low"
```

```python
def generate_invoice(order, products_list, customers_list):

    # Get customer name
    customer_name = get_customer_name(
        order["Customer_ID"],
        customers_list
    )

    # Get product price
    product_price = get_product_price(
        order["Product_ID"],
        products_list
    )

    # Find product name
    product_name = ""

    for product in products_list:
        if product["Product_ID"] == order["Product_ID"]:
            product_name = product["Name"]
            break

    # Calculate order value
    order_value = calculate_order_value(
        product_price,
        order["Quantity"]
    )

    # Return formatted invoice
    return f"""
Order ID: {order["Order_ID"]}
Customer: {customer_name}
Product: {product_name}
Quantity: {order["Quantity"]}
Order Value: {order_value}
Status: {order["Status"]}
"""
```

```python
def calculate_total_revenue(orders_list, products_list):

    total_revenue = 0

    for order in orders_list:

        # Skip cancelled orders
        if order["Status"] == "Cancelled":
            continue

        price = get_product_price(
            order["Product_ID"],
            products_list
        )

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        total_revenue += order_value

    return total_revenue
```

```python
def top_customer_by_spend(orders_list, products_list, customers_list):

    customer_spending = {}

    for order in orders_list:

        # Only Delivered orders count
        if order["Status"] != "Delivered":
            continue

        customer_id = order["Customer_ID"]

        price = get_product_price(
            order["Product_ID"],
            products_list
        )

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        # Add spending for each customer
        if customer_id in customer_spending:
            customer_spending[customer_id] += order_value
        else:
            customer_spending[customer_id] = order_value


    # Find customer with highest spending
    top_customer_id = None
    highest_spending = 0

    for customer_id in customer_spending:

        if customer_spending[customer_id] > highest_spending:
            highest_spending = customer_spending[customer_id]
            top_customer_id = customer_id


    return get_customer_name(
        top_customer_id,
        customers_list
    )
```

```python
def apply_offer(price, *discounts):

    for discount in discounts:
        price = price - (price * discount / 100)

    return price
```

```python
print("=" * 50)
print("       URBANCART BUSINESS SUMMARY REPORT")
print("=" * 50)


# -----------------------------------
# 1. Product and category information
# -----------------------------------

total_products = len(products)

categories = set()

for product in products:
    categories.add(product["Category"])

total_categories = len(categories)


# -----------------------------------
# 2. Customer membership information
# -----------------------------------

total_customers = len(customers)

premium_count = 0
regular_count = 0

for customer in customers:
    if customer["Membership"] == "Premium":
        premium_count += 1
    else:
        regular_count += 1


# -----------------------------------
# 3. Order status information
# -----------------------------------

total_orders = len(orders)

delivered_count = 0
pending_count = 0
cancelled_count = 0

for order in orders:

    if order["Status"] == "Delivered":
        delivered_count += 1

    elif order["Status"] == "Pending":
        pending_count += 1

    elif order["Status"] == "Cancelled":
        cancelled_count += 1


# -----------------------------------
# 4. Total Revenue
# Delivered orders only
# -----------------------------------

total_revenue = calculate_total_revenue(orders, products)


# -----------------------------------
# 5. Average Order Value
# Delivered orders only
# -----------------------------------

average_order_value = total_revenue / delivered_count


# -----------------------------------
# 6. Top Customer
# -----------------------------------

top_customer = top_customer_by_spend(
    orders,
    products,
    customers
)


# -----------------------------------
# 7. Out-of-stock products
# -----------------------------------

out_of_stock_count = 0

for product in products:
    if product["Stock"] == 0:
        out_of_stock_count += 1


# -----------------------------------
# 8. Most popular product category
# Delivered orders only
# -----------------------------------

category_count = {}

for order in orders:

    # Only delivered orders count
    if order["Status"] != "Delivered":
        continue

    # Find the matching product
    for product in products:

        if order["Product_ID"] == product["Product_ID"]:

            category = product["Category"]

            # Count the category
            if category in category_count:
                category_count[category] += 1
            else:
                category_count[category] = 1

            break


# Find the category with the highest count

most_popular_category = None
highest_count = 0

for category in category_count:

    if category_count[category] > highest_count:
        highest_count = category_count[category]
        most_popular_category = category


# -----------------------------------
# FINAL REPORT
# -----------------------------------

print("\nPRODUCT SUMMARY")
print("-" * 50)
print("Total Products:", total_products)
print("Distinct Categories:", total_categories)


print("\nCUSTOMER SUMMARY")
print("-" * 50)
print("Total Customers:", total_customers)
print("Premium Members:", premium_count)
print("Regular Members:", regular_count)


print("\nORDER SUMMARY")
print("-" * 50)
print("Total Orders:", total_orders)
print("Delivered Orders:", delivered_count)
print("Pending Orders:", pending_count)
print("Cancelled Orders:", cancelled_count)


print("\nSALES SUMMARY")
print("-" * 50)
print("Total Revenue:", total_revenue)
print("Average Order Value:", round(average_order_value, 2))
print("Top Customer:", top_customer)


print("\nINVENTORY SUMMARY")
print("-" * 50)
print("Out-of-Stock Products:", out_of_stock_count)


print("\nPOPULAR CATEGORY")
print("-" * 50)
print("Most Popular Category:", most_popular_category)
print("Number of Delivered Orders:", highest_count)


print("\n" + "=" * 50)
print("              END OF REPORT")
print("=" * 50)
```
